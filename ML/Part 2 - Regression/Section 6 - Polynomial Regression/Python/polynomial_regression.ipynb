# Polynomial Regression

## Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```python
dataset = pd.read_csv('Position_Salaries.csv')
X = dataset.iloc[:, 1:-1].values #Matrix of features, here we are only going to use the second column because the first column is just the position name which is not a numerical value, and the last column is the target variable
y = dataset.iloc[:, -1].values #Target variable
```

## Training the Linear Regression model on the whole dataset

```python
#Here we are going to train the Linear Regression model first to see the difference between the two models
from sklearn.linear_model import LinearRegression
lin_regressor = LinearRegression()
lin_regressor.fit(X, y)
```

## Training the Polynomial Regression model on the whole dataset

```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=4)
poly.fit_transform (X) #This will create a new matrix of features with the polynomial features, for example if we have a feature x, it will create a new feature x^2
X_poly = poly.fit_transform(X)

from sklearn.linear_model import LinearRegression
lin_regressor2 = LinearRegression()
lin_regressor2.fit(X_poly, y ) #Here we trained this linear regression model with the polynomial features
```

## Visualising the Linear Regression results

```python
plt.scatter(X, y, color = 'red')
plt.plot(X, lin_regressor.predict(X), color = 'blue') #This is the linear regression model
plt.title('Truth or Bluff (Linear Regression)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()

#notice that the linear regression model is not a good fit for the data
```

## Visualising the Polynomial Regression results

```python
plt.scatter(X, y, color = 'red')
plt.plot(X, lin_regressor2.predict(X_poly), color = 'blue') #This is the polynomial regression model
plt.title('Truth or Bluff (Polynomial Regression of degree 2 )')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()

#Notice that the polynomial regression model is a better fit than the linear regression model, but it is still not accurate.
```

## Visualising the Polynomial Regression results (for higher resolution and smoother curve)

```python
plt.scatter(X, y, color = 'red')
plt.plot(X, lin_regressor2.predict(X_poly), color = 'blue') #This is the polynomial regression model
plt.title('Truth or Bluff (Polynomial Regression of degree 4 )')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()

#Notice that after changing the degree of the polynimal, the model is a perfect fit for the data, and it is able to predict the salary of the employee with a high degree of accuracy.

#This could be considered overfitting, but we won't worry about that for now.
```

## Predicting a new result with Linear Regression

```python
#We are now going to predict the salary of an employee with a position level of 6.5 using both models

lin_regressor.predict([[6.5]]) #This is the prediction using the linear regression model
#Clearly, the prediction is not very accurate
```

## Predicting a new result with Polynomial Regression

```python
#We are now going to predict the salary with the polynomial regression model

lin_regressor2.predict(poly.fit_transform([[6.5]])) 

#Notice that this is much more accurate than the linear regression model
```
