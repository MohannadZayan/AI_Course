# Simple Linear Regression

## Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```python
dataset = pd.read_csv('Salary_Data.csv')

X = dataset.iloc[: , :-1].values # matrix of feautures, [: , :-1] means all rows and all columns except the last one
y = dataset.iloc[:, -1].values # vector of dependent variable, [: , -1] means all rows and only the last column
```

## Splitting the dataset into the Training set and Test set

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

## Training the Simple Linear Regression model on the Training set

```python
from sklearn.linear_model import LinearRegression #importing the linear regression model (or class) from the linear_model module
regressor = LinearRegression() #creating an object of the class LinearRegression
regressor.fit(x_train, y_train) #fitting the model to the training data
```

## Predicting the Test set results

```python
y_pred = regressor.predict (x_test) #predicting the test set results
```

## Visualising the Training set results

```python
plt.scatter (x_train, y_train, color = 'red')  #plotting the training (real) data points
plt.plot (x_train, regressor.predict(x_train), color = 'blue')  #plotting the predicted data points from the training set
plt.title ('Salary vs Experience (Training set)')
plt.xlabel ('Years of Experience')
plt.ylabel ('Salary')
plt.show ()
```

## Visualising the Test set results

```python
plt.scatter (x_test, y_test, color = 'red')  #plotting the training (real) data points
plt.plot (x_test, regressor.predict(x_test), color = 'blue')  #plotting the predicted data points from the test set
plt.title ('Salary vs Experience (Test set)')
plt.xlabel ('Years of Experience')
plt.ylabel ('Salary')
plt.show ()
```
