# Data Preprocessing Tools

## Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```python
dataset = pd.read_csv("Data.csv")
x = dataset.iloc [: , :-1].values #MATRIX OF FEATURES, AKA THE VARIABLES USED FOR PREDICTION
y = dataset.iloc [: , -1].values #VECTOR OF DEPENDENT VARIABLE, AKA THE VARIABLE TO BE PREDICTED
```

```python
x
```

```python
y
```

## Taking care of missing data

```python
#replacing the missing values with the mean of the column:

from sklearn.impute import SimpleImputer

imputer = SimpleImputer(missing_values=np.nan, strategy='mean') #identifies the missing data entry and states the strategy to be used in replacing them
imputer.fit (x[: , 1:3]) #tells the imputer where the missing values are in the data, here it's the numerical values in the first and second columns
x [: , 1:3] = imputer.transform (x [: , 1:3]) #here it actually updates the matrix x with the imputed values
```

```python
print (x)
```

## Encoding categorical data

### Encoding the Independent Variable

```python
#Here we have to change the countries names to be numerical values but we have to let the model know that the countries are not ordered

#We can use one hot encoding to turn the country column into 3 columns because there are only 3 countries ; 3 categories

#We will also replace the values in the "Purchased" column with zeroes and ones

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

#Creating an object of the ColumnTransformer class:
#The 2 transformer arguments will be what kind of transformation we want to apply to the data and on which indexes of the columns, and the second argument is the remainder, which will stay the same.
ct = ColumnTransformer(transformers = [('encoder', OneHotEncoder(), [0])], remainder = 'passthrough')
x = np.array(ct.fit_transform(x)) #Connecting the transformer to the matrix of features 'x'
```

```python
print (x)
```

### Encoding the Dependent Variable

```python
#Here we are going to instead encode the dependant variable, AKA purchased, as a binary variable.

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y = le.fit_transform(y)
```

```python
y
```

## Splitting the dataset into the Training set and Test set

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=1)
```

## Feature Scaling

```python
#Feauture scaling MUST be done AFTER splitting the dataset into training and test sets, so that the test set is not used during the scaling of the training set.
#It's done to level the playing field between the features, so that one feature doesn't dominate the others because of its scale.

from sklearn.preprocessing import StandardScaler

#Standardizing the data
sc = StandardScaler()

#Before we scale, notice that some values in the matrix of features are 0 and 1, which are already standardized, so we will not scale those values, but we will scale the other values in the matrix.

X_train[:, 3:] = sc.fit_transform(X_train[:, 3:]) #We fit the scaler to the training set and then transform it, so that the training set is scaled.
X_test[:, 3:] = sc.transform(X_test[:, 3:]) #We transform the test set, so that the test set is scaled, we dont fit it because we already fit the training set.
```
